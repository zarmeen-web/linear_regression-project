```python
X = df["YearsExperience"].values
y = df["Salary"].values
```

**gradient** **descent**

```python
import numpy as np
```

```python
m = 0
b = 0

learning_rate = 0.01
epochs = 10000

X_mean = X.mean()
X_std = X.std()

X_scaled = (X - X_mean) / X_std

for i in range(epochs):

    y_pred = m * X_scaled + b

    dm = (2 / len(X)) * np.sum((y_pred - y) *X_scaled)
    db = (2 / len(X)) * np.sum(y_pred - y)

    m = m - learning_rate * dm
    b = b - learning_rate * db
```

```python
m_original= m / X_std
b_original= b - m_original * X_mean

print("Coefficient:", m_original)
print("Intercept:", b_original)
```

```python
sal=(m_original*5.6)+b_original
sal
```
